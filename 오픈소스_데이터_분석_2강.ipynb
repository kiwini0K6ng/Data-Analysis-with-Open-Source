## 2-1 리스트 슬라이싱

```python
numbers = [10, 20, 30, 40, 50, 60, 70]
# 첫 4개의 원소 출력
first_four = numbers[:4]
print(first_four)
# 뒤 3개의 원소를 출력
last_three = numbers[-3:]
print(last_three)
```

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

# 홀수번째 항목 출력
every_other_list = numbers[::2]
print(every_other_list)

# 역순 출력
reverse_list = numbers[::-1]
print(reverse_list)
```

## 2-2 리스트 컴프리헨션

```python
numbers = [ ]

# 0 ~ 4 포함 리스트 생성
#for i in range(5):
#   numbers.append(i)
numbers =[i for i in range(5)]
print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
squares = [ ]

# numbers의 모든 항목의 값을 제곱한 리스트
#for num in numbers:
#  squares.append(num ** 2)
squares = [num**2 for num in numbers]
print(squares)
```

## 2-3 문자열 형식화

```python
name = '홍길동'
age = 30
salary = 3500000
tax_rate = 0.1

# C 스타일 문자열 포매팅
# 이름, 나이, 월급 출력
basic_format='%s,%d,%d' % (name,age,int(salary*(1-tax_rate)))
# 직원이름, 나이, 세후 월급 출력
index_format = '직원이름:%s, 나이:%d, 세후월급:%d' % (name,age,int(salary*(1-tax_rate)))

# 직원이름, 나이, 월급, 세금, 실수령액 추력


print(basic_format)
print(index_format)
```

```python
name = '홍길동'
age = 30
salary = 3500000
tax_rate = 0.1

# str.format 문자열 포매팅
# 이름, 나이, 월급 출력
basic_format='이름:{},나이:{},월급'.format(name,age,salary)

# 직원이름, 나이, 세후 월급 출력
index_format = '''직원이름 : {0}, 나이 : {1} 세후 월급 : {2:,}'''.format(name,age,int(salary*(1-tax_rate)))
# 직원이름, 나이, 월급, 세금, 실수령액 추력
keyword_format ='''직원이름 : {employee}, 나이 : {age}, 월급: {salary:,},실수령액{real:,}'''.format(employee=name,age=age,salary=salary,
                     real=int(salary*(1-tax_rate)))
print(basic_format)
print(index_format)
print(keyword_format)
```
